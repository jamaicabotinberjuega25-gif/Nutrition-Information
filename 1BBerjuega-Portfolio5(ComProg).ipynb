**Program: Nutrition Information**  
&emsp; A string in computer programming is defined as an object which can process natural human language. In most cases, this would include names, words, and sentences. Data from the real world is often not formatted properly, with words such as "ml" or "g" in between. String objects can be employed in order to address this issue, allowing computers to extract numeric values from the information and calculate mathematically.

&emsp;In this program,  Python’s string manipulation capabilities will be employed to extract useful information from the Nestea Green Tea nutrition label, by removing the text elements and converting the numerical digits into actual numbers so that the total mass of the nutrients in the 500mL bottle can be calculated using its 230mL serving size.

* The nutrition facts label on a Nestea Green Tea bottle indicates the amount of nutrients per one serving size, which is 230mL. In reality, the bottle's capacity is 500mL. Since the information on the nutrition facts label includes texts (like "20.2 g" and "30.9 mg"), the computer will not be able to perform calculations with these texts. It is necessary to develop a program that will eliminate these texts, convert the information into numeric values, and find out how many carbohydrates, sugars, and sodium are there in 500mL of the beverage.

![image](https://www.image2url.com/r2/default/images/1779254396411-20c922ec-4f17-4227-99f4-5f53265a3932.jpg)

```python
# Raw Data Input 
# The data from the label is stored as strings mixed with units.
label_serving_size = "230ml"
label_carbs = "20.2 g"
label_sugars = "19.8 g"
label_sodium = "30.9 mg"

target_volume = 500  # Total bottle volume in mL

#  String Parsing and Cleaning ---
# We use string slicing [:-2] to remove the last two characters ("ml", " g", "mg")
# and .strip() to remove any blank spaces left behind.
serving_clean = label_serving_size[:-2].strip()
carbs_clean = label_carbs[:-2].strip()
sugars_clean = label_sugars[:-2].strip()
sodium_clean = label_sodium[:-2].strip()

# 3. Converting Strings to Numbers
# We convert the clean strings into float numbers so we can do math.
serving_num = float(serving_clean)
carbs_num = float(carbs_clean)
sugars_num = float(sugars_clean)
sodium_num = float(sodium_clean)

# 4. Calculation
# Finding the multiplier by dividing the total bottle size by the serving size.
scale_factor = target_volume / serving_num

# Multiplying the clean numbers by our scale factor
total_carbs = carbs_num * scale_factor
total_sugars = sugars_num * scale_factor
total_sodium = sodium_num * scale_factor

# 5. Final Output Report
report = f"""

Nestea Green Tea Nutrient Scaling (500mL Bottle)

Calculated Multiplier: {scale_factor:.4f}x

Total Nutrients in 500mL:
• Total Carbohydrates : {total_carbs:.2f} g
• Total Sugars        : {total_sugars:.2f} g
• Total Sodium        : {total_sodium:.2f} mg
"""

print(report)
```
